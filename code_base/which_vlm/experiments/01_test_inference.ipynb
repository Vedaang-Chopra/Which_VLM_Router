```python
CONFIG_PATH = (Path.cwd().parent / "configs" / "inference_vlm.yaml").resolve()

client = WhichVLMClient.from_yaml(CONFIG_PATH)

print("All models: ", client.list_models())
print("LLM models:", client.list_llm_models())
print("VLM models:", client.list_vlm_models())
```

### LLM Only Test

```python
prompt = "Explain what a router is in one simple sentence."

llm_results = client.llm.run_single(prompt, models="all")

for name, out in llm_results.items():
    print("=== LLM:", name, "===")
    if out.get("ok"):
        print(out["response_text"])
        print("Latency (ms):", out["latency_ms"])
        print("Estimated cost:", out["est_cost"])
    else:
        print("ERROR:", out.get("error"))
```

```python
prompts = [
    "1+1=",
    "2+2=",
    "Describe what a router does in 10 words.",
]

llm_model = client.list_llm_models()[0]  # first LLM
batch_results = client.llm.run_batch(prompts, model=llm_model)

for p, out in zip(prompts, batch_results):
    print("PROMPT:", p)
    if out.get("ok"):
        print(" ->", out["response_text"])
    else:
        print("ERROR:", out.get("error"))
    print()
```

### VLM Test

```python
path_to_image = "/Users/vedaangchopra/all_data/complete_technical_work/all_projects_implemented/Which_VLM_Router/dataset/cartoon_describe.png"  # 
```

```python
vlm_results = client.vlm.run_image(
    image=path_to_image,
    text="What is happening in this image?",
    models="all",
)

for name, out in vlm_results.items():
    print("=== VLM:", name, "===")
    if out.get("ok"):
        print(out["response_text"])
        print("Latency (ms):", out["latency_ms"])
    else:
        print("ERROR:", out.get("error"))
```

```python
desc_results = client.vlm.describe_image(path_to_image, models="all")

for name, text in desc_results.items():
    print("=== VLM describe:", name, "===")
    print(text)
```

```python
examples = [
    {"image": "cat.png", "text": "What animal is this?"},
    {"image": "dog.png", "text": "What animal is this?"},
]

vlm_batch_results = client.vlm.run_batch(examples, models="all")

for model_name, outputs in vlm_batch_results.items():
    print("=== Model:", model_name, "===")
    for i, out in enumerate(outputs):
        if out.get("ok"):
            print(f"Example {i} ->", out["response_text"])
        else:
            print(f"Example {i} ERROR ->", out.get("error"))
    print()
```
